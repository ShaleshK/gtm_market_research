```python
import os
from pathlib import Path
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, Process, Flow
from crewai.project import CrewBase, agent, crew, task
from langchain_openai import AzureChatOpenAI

# 1. Traversal Path Math: Fetch keys from one directory level up
subfolder_path = Path(os.getcwd())
main_folder_dotenv = subfolder_path.parent / ".env"
load_dotenv(dotenv_path=main_folder_dotenv)

# 2. Bind your production Azure OpenAI Endpoint engine
azure_llm = AzureChatOpenAI(
    azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
    api_version=os.getenv("AZURE_OPENAI_API_VERSION", "2024-02-15-preview"),
    deployment_name=os.getenv("AZURE_OPENAI_CHAT_DEPLOYMENT", "vt-agi-chat")
)

print("✅ Azure OpenAI Engine Authenticated. Capstone pipeline armed!")
```

```python
# 1. Head Planner Agent (The Orchestrator and Documenter)
head_planner = Agent(
    role="Head GTM Planner and Orchestrator",
    goal="Coordinate the research workflow streams, synthesize specialist findings, and compile the final GTM plan.",
    backstory="You are an elite corporate systems operator. You digest raw project briefs, outline exact research questions, and manage the execution steps of your sub-agents.",
    llm=azure_llm,
    verbose=True
)

# 2. Research Agent (The Finder)
research_agent = Agent(
    role="Primary Market Research Scientist",
    goal="Gather hard empirical evidence, retrieve competitor pricing data, and extract verified URL citations.",
    backstory="You are a data-mining specialist. You track tracking sources, ignore unverified blogs, and ensure 100% of data claims are backed by premium primary links.",
    llm=azure_llm,
    verbose=True
)

# 3. Analyst Agent (The Sense-Maker)
analyst_agent = Agent(
    role="Senior Competitive Intelligence Analyst",
    goal="Convert raw web snippets into structured strategic arrays, competitive tables, and analytical frameworks.",
    backstory="You specialize in strategic modeling. You ingest raw research dumps and map them out into clean SWOT, 4P, and 7P matrices to uncover industry vulnerabilities.",
    llm=azure_llm,
    verbose=True
)

# 4. Strategy Agent (The Planner)
strategy_agent = Agent(
    role="Principal Go-To-Market Strategist",
    goal="Formulate high-differentiation GTM playbooks, ideal customer profiles (ICPs), and launch timeline maps.",
    backstory="You are a world-class growth architect. You take competitive intelligence grids and translate them into actionable, step-by-step launch plans.",
    llm=azure_llm,
    verbose=True
)
```

```python
# Define the core project challenge parameters
project_brief = "Launching a disruptive multi-agent validation runtime software service for mid-market B2B SaaS engineering teams."

# Task 1: Head Planner Scoping Pass
task_plan = Task(
    description=f"Analyze this project brief: '{project_brief}'. Generate a structured multi-point market research map outlining 5 core research target nodes.",
    expected_output="A structured markdown outline containing 5 strategic research question clusters.",
    agent=head_planner
)

# Task 2: Research Gathering Pass with Citations
task_research = Task(
    description="Execute deep web searching across the scoped nodes. Retrieve primary source URL links, market evidence statistics, and competitor data models.",
    expected_output="A structured JSON data matrix mapping out findings directly to verified source link variables with 0% broken links.",
    agent=research_agent
)

# Task 3: Analyst Matrix Construction Pass
task_analysis = Task(
    description="Analyze the JSON research outputs. Generate a structured competitive landscape matrix table alongside a comprehensive SWOT and 4P marketing analysis.",
    expected_output="A clean markdown document containing formatted competitive comparison tables and full qualitative SWOT/4P text grids.",
    agent=analyst_agent
)

# Task 4: Strategist Final Blueprint Pass
task_gtm_draft = Task(
    description="Ingest the competitive matrices and compile the finalized Go-To-Market blueprint document. Structure explicit Ideal Customer Profiles (ICPs), core value propositions, and a detailed 90-day execution launch plan.",
    expected_output="A fully articulated, publication-ready GTM strategy document containing distinct headers, ICP tables, and launch milestones.",
    agent=strategy_agent
)
```

```python
# Assemble your agents and chained tasks into an automated execution crew
gtm_crew = Crew(
    agents=[head_planner, research_agent, analyst_agent, strategy_agent],
    tasks=[task_plan, task_research, task_analysis, task_gtm_draft],
    process=Process.sequential,  # Enforces clean round-robin task handoffs
    verbose=True
)

print("🚀 Launching CrewAI Multi-Agent GTM Execution Loop...")
final_gtm_blueprint = gtm_crew.kickoff()

# Display the finalized master document content block output right inside your notebook layout!
print("\n🏆 FINAL COMPLETED GTM BLUEPRINT:\n")
print(final_gtm_blueprint)
```
